# Customer Age EDA

Exploratory analysis of customer age distribution.

```python
import pandas as pd
import sqlalchemy

engine = sqlalchemy.create_engine('postgresql://localhost/analytics')

df = pd.read_sql("""
    SELECT
        customer_id,
        customer_age,
        CASE
            WHEN customer_age < 25 THEN 'young'
            WHEN customer_age < 60 THEN 'middle'
            ELSE 'senior'
        END as age_bucket
    FROM dim_customers
    WHERE customer_age IS NOT NULL
""", engine)

print(df.describe())
```
